```python
def impute_q6(df):
    q6_cols = ['q61','q62','q63', 'q64', 'q65']
    for index, row in df.iterrows():
        nan_count = row[q6_cols].isnull().sum()
        if nan_count == 5:
            df.loc[index, q8_cols] = 0  # Impute with 0 if all 5 columns are NaN
        elif nan_count > 0:
            numeric_row = pd.to_numeric(row[q6_cols], errors='coerce')
            non_nan_sum = numeric_row.sum()

            if non_nan_sum > 100:  # if non null sum > 100, we need to impute null with 0 first
            # and will decide how to deal with them in below section (otherwise we will be imputing with negative values)
                df.loc[index, q6_cols] = df.loc[index, q6_cols].fillna(0)
            else:
                # Impute NaN values using (100 - non_nan_sum) / nan_count
                value_to_impute = (100 - non_nan_sum) / nan_count
                for col in q6_cols:
                    if pd.isna(df.loc[index, col]):
                        df.loc[index, col] = value_to_impute
    return df
```

```python
q8_cols = ['q81','q82','q83','q84', 'q85']
```

```python
def impute_q8(df):
    q8_cols = ['q81', 'q82', 'q83', 'q84', 'q85']
    for index, row in df.iterrows():
        nan_count = row[q8_cols].isnull().sum()
        if nan_count == 5:
            df.loc[index, q8_cols] = 0  # Impute with 0 if all 5 columns are NaN
        elif nan_count > 0:
            numeric_row = pd.to_numeric(row[q8_cols], errors='coerce')
            non_nan_sum = numeric_row.sum()

            if non_nan_sum > 100:  # if non null sum > 100, we need to impute null with 0 first
            # and will drop these rows in below section (otherwise we will be imputing with negative values)
                df.loc[index, q8_cols] = df.loc[index, q8_cols].fillna(0)
            else:
                # Impute NaN values using (100 - non_nan_sum) / nan_count
                value_to_impute = (100 - non_nan_sum) / nan_count
                for col in q8_cols:
                    if pd.isna(df.loc[index, col]):
                        df.loc[index, col] = value_to_impute
    return df
```

Explanation:

We impute missing values in the `q8_cols` (q81-q85) based on these rules:

1. **All 5 null:** If all five columns in `q8_cols` are null for a row, they are imputed with 0. (representing blank data) --> responders might just leave them as blank for some reasons, but we don't have more information to impute.

2. **Some null, Sum > 100:** If some, but not all, `q8_cols` are null and the sum of the non-null values is greater than 100, the null are imputed with 0. These rows will be dropped later in next section (cuz it was small portion anyway)

3. **Some null, Sum <= 100:** If some `q8_cols` are null and the sum of non-null values is less than or equal to 100, the null values are imputed with a value calculated to make the sum of the five columns equal to 100.

```python
from scipy.stats import norm

def logit_pvalue(model, x):
    # Convert x to numeric values
    x = x.astype(float)  # Add this line to convert x to numeric
    p = model.predict_proba(x)
    n = len(p)
    m = len(model.coef_[0]) + 1
    coefs = np.concatenate([model.intercept_, model.coef_[0]])
    x_full = np.matrix(np.insert(np.array(x), 0, 1, axis = 1))
    ans = np.zeros((m, m))
    for i in range(n):
        ans = ans + np.dot(np.transpose(x_full[i, :]), x_full[i, :]) * p[i,1] * p[i, 0]
    vcov = np.linalg.inv(np.matrix(ans))
    se = np.sqrt(np.diag(vcov))
    t =  coefs/se
    p = (1 - norm.cdf(abs(t))) * 2
    return p



print(logit_pvalue(logistic_model, X_train))
```
